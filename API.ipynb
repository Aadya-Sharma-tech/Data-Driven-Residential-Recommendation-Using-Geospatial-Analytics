```python
def count_amenity(lat, lon, amenity):

    params = {
        "ll": f"{lat},{lon}",
        "query": amenity,
        "radius": 1500,
        "limit": 50
    }

    response = requests.get(
        url,
        headers=headers,
        params=params
    )

    data = response.json()

    return len(data.get('results', []))
```

```python
count_amenity(test_lat, test_lon, "restaurant")
```

```python
locations['restaurants'] = locations.apply(
    lambda row: count_amenity(
        row['latitude'],
        row['longitude'],
        'restaurant'
    ),
    axis=1
)
```

```python
locations['grocery_stores'] = locations.apply(
    lambda row: count_amenity(
        row['latitude'],
        row['longitude'],
        'grocery store'
    ),
    axis=1
)
```

```python
locations['gyms'] = locations.apply(
    lambda row: count_amenity(
        row['latitude'],
        row['longitude'],
        'gym'
    ),
    axis=1
)
```

```python
locations.head()
```

```python
locations[['restaurants', 'grocery_stores', 'gyms']].describe()
```

```python
location_features = locations[[
    'restaurants',
    'grocery_stores',
    'gyms'
]].copy()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_location = scaler.fit_transform(location_features)
```

```python
optimal_k = 2
```

```python
from sklearn.cluster import KMeans

kmeans_location = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

locations['Cluster'] = kmeans_location.fit_predict(X_location)
```

```python
locations.head()
```

```python
locations.groupby('Cluster')[
    ['restaurants', 'grocery_stores', 'gyms']
].mean()
```
